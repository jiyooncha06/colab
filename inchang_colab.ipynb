구구단 만들기 코드를 for문을 써서 만들어줘. 몇단을 할지는 입력을 받아서 하게해줘.

```python
# 2단부터 9단까지 전체 구구단을 출력합니다.
for dan in range(2, 10):
    print(f"\n=== {dan}단 ===")
    for i in range(1, 10):
        print(f"{dan} x {i} = {dan * i}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# 별의 꼭짓점 좌표를 구하는 함수
def get_star_coords(cx, cy, r_outer, r_inner):
    angles = np.linspace(0, 2 * np.pi, 11) + np.pi/2
    x = []
    y = []
    for i, angle in enumerate(angles):
        r = r_outer if i % 2 == 0 else r_inner
        x.append(cx + r * np.cos(angle))
        y.append(cy + r * np.sin(angle))
    return np.array(x), np.array(y)

# 애니메이션을 위한 피겨 설정
fig, ax = plt.subplots(figsize=(8, 6), facecolor='black')
ax.set_facecolor('black')
ax.set_xlim(0, 100)
ax.set_ylim(0, 100)
ax.axis('off')

# 배경에 작은 고정된 별들 생성
np.random.seed(42)
background_x = np.random.rand(50) * 100
background_y = np.random.rand(50) * 100
ax.scatter(background_x, background_y, s=2, color='white', alpha=0.5)

# 떨어지는 별똥별 오브젝트 생성 (처음에는 화면 밖에 위치)
star_patch, = ax.plot([], [], color='gold', linewidth=1.5)
star_fill = ax.fill([], [], color='yellow', alpha=0.8)[0]
trail_patch, = ax.plot([], [], color='white', alpha=0.6, linewidth=2)  # 별똥별 꼬리

# 프레임 수 및 궤적 정의 (왼쪽 위 -> 오른쪽 아래)
num_frames = 60
start_x, start_y = 10, 90
end_x, end_y = 90, 10

path_x = np.linspace(start_x, end_x, num_frames)
path_y = np.linspace(start_y, end_y, num_frames)

def update(frame):
    cx, cy = path_x[frame], path_y[frame]

    # 별 좌표 계산 (바깥 반지름 3, 안쪽 반지름 1.2)
    sx, sy = get_star_coords(cx, cy, 3, 1.2)
    star_patch.set_data(sx, sy)

    # 별 채우기 업데이트
    star_vertices = np.column_stack((sx, sy))
    star_fill.set_xy(star_vertices)

    # 별똥별 꼬리 생성 (이전 프레임의 위치들을 이어 붙임)
    trail_len = 8
    start_idx = max(0, frame - trail_len)
    trail_x = path_x[start_idx:frame+1]
    trail_y = path_y[start_idx:frame+1]
    trail_patch.set_data(trail_x, trail_y)

    return star_patch, star_fill, trail_patch

# 애니메이션 생성
anim = FuncAnimation(fig, update, frames=num_frames, interval=50, blit=True)
plt.close()

# Colab에서 애니메이션을 볼 수 있도록 HTML 변환
HTML(anim.to_jshtml())
```
